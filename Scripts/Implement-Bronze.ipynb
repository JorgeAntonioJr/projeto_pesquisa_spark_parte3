# Create Dataframe from db(.csv) files

```python
caminho_landing = '/Volumes/workspace/landing/dados'

df_teacher   = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/teacher.csv")
df_classes   = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/classes.csv")
df_students   = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/students.csv")
df_subjects   = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/subjects.csv")
df_grades   = spark.read.option("infeschema", "true").option("header", "true").csv(f"{caminho_landing}/grades.csv")
```

Adiciona uma nova coluna (metadado) de data e hora de processamento e nome do arquivo de origem.

# Adds time and file name as columns

```python
from pyspark.sql.functions import current_timestamp, lit


df_teacher = df_teacher.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("teacher.csv"))
df_classes = df_classes.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("classes.csv"))
df_students = df_students.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("students.csv"))
df_subjects = df_subjects.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("subjects.csv"))
df_grades = df_grades.withColumn("data_hora_bronze", current_timestamp()).withColumn("nome_arquivo", lit("grades.csv"))
```

# Save Delta Tables

```python
df_subjects.write.format('delta').mode("overwrite").saveAsTable("bronze.subjects")
df_classes.write.format('delta').mode("overwrite").saveAsTable("bronze.classes")
df_students.write.format('delta').mode("overwrite").saveAsTable("bronze.students")
df_teacher.write.format('delta').mode("overwrite").saveAsTable("bronze.teacher")
df_grades.write.format('delta').mode("overwrite").saveAsTable("bronze.grades")
```
